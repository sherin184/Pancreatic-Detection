# Dataset analysis for pancreatic tumour classification

This notebook inspects the dataset and saves summary figures to `results/figures/`.
The testing set remains untouched and is used only for the final evaluation stage.

```python
from pathlib import Path
import matplotlib
matplotlib.use('Agg')
import matplotlib.pyplot as plt

from src.dataset import CLASS_NAMES, inspect_dataset

summary = inspect_dataset()
print('Training class distribution:')
for cls, count in summary['training'].items():
    print(f'  {cls}: {count}')
print('\nTesting class distribution:')
for cls, count in summary['testing'].items():
    print(f'  {cls}: {count}')
print(f'\nImage formats: {dict(summary[formats])}')
print(f'Corrupted images: {summary[corrupted_images]}')
if summary['average_dimensions'] is not None:
    print(f"Average image dimensions: {summary['average_dimensions']['width']:.1f} x {summary['average_dimensions']['height']:.1f}")
print(f"Minimum pixel area: {summary['min_size']}")
print(f"Maximum pixel area: {summary['max_size']}")
```

```python
from src.dataset import DATASET_ROOT, CLASS_NAMES
from PIL import Image
from pathlib import Path

summary = inspect_dataset(DATASET_ROOT)
training_counts = [summary['training'][cls] for cls in CLASS_NAMES]
testing_counts = [summary['testing'][cls] for cls in CLASS_NAMES]

plt.figure(figsize=(10, 4))
plt.subplot(1, 2, 1)
plt.bar(CLASS_NAMES, training_counts, color=['tab:blue', 'tab:orange'])
plt.title('Training Class Distribution')
plt.ylabel('Count')
plt.xticks(rotation=15)

plt.subplot(1, 2, 2)
plt.bar(CLASS_NAMES, testing_counts, color=['tab:green', 'tab:red'])
plt.title('Testing Class Distribution')
plt.ylabel('Count')
plt.xticks(rotation=15)
plt.tight_layout()
plt.savefig(Path('results/figures') / 'dataset_distribution.png', dpi=200)
plt.close()
```

```python
from pathlib import Path
from PIL import Image
import matplotlib
matplotlib.use('Agg')
import matplotlib.pyplot as plt

root = Path('dataset')
sample_paths = []
for class_name in ['Normal', 'Pancreatic Tumour']:
    class_dir = root / 'Training' / class_name
    files = sorted(class_dir.glob('*'))
    sample_paths.extend(files[:3])

fig, axes = plt.subplots(2, 3, figsize=(12, 6))
for ax, path in zip(axes.ravel(), sample_paths):
    image = Image.open(path).convert('RGB')
    ax.imshow(image)
    ax.set_title(path.parent.name)
    ax.axis('off')

for idx in range(len(sample_paths), len(axes.ravel())):
    axes.ravel()[idx].axis('off')

plt.tight_layout()
plt.savefig(Path('results/figures') / 'sample_images.png', dpi=200)
plt.close()
```
